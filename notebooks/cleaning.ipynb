```python
import pandas as pd
```

```python
df = pd.read_csv('../data/Synthetic_Transportation_Dataset_Expanded_v2.csv')
print(f"Loaded : {len(df):,} rows, {len(df.columns)} columns")
print()
df.columns
```

```python
df.head()
```

```python
df.info()
```

```python
print(f"Missing values per column:")
print(df.isnull().sum())
print(f"\nDuplicate rows: {df.duplicated().sum()}")
print(f"Timestamp dtype: {df['Timestamp'].dtype}")
print(f"Sample timestamps: {df['Timestamp'].head(3).tolist()}")
print()
```

```python
df['Timestamp'] = pd.to_datetime(df['Timestamp'], format='%d-%m-%Y %H:%M')
print(f"Timestamp parsed -> dtype: {df['Timestamp'].dtype}")
```

```python
df['Year']      =  df['Timestamp'].dt.year
df['Month']     =  df['Timestamp'].dt.month
df['Day']       =  df['Timestamp'].dt.day
df['Hour']      =  df['Timestamp'].dt.hour
df['DayOfWeek'] =  df['Timestamp'].dt.day_name()
print("Extracted : Year, Month, Day, Hour, DayOfWeek")
```

```python
df = df.rename(columns={
    'temprature' : 'Temprature',
    'humidity': 'Humidity',
    'Avg_Speed(km/h)': 'Avg_Speed_kmh',
    'Visibility(m)': 'Visibility_m'
    })
```

```python
df['Accident_Severity'] = df['Accident_Severity'].replace('-', 'None')
```

```python
df['Accident_Flag'] = df['Accident_Occurred'].map({'Yes': 1, 'No': 0})
df['Alert_Flag']    = df['Alert_Generated'].map({'Yes': 1, 'No': 0})
```

```python
print(f"Shape: {df.shape}")
print(f"\nMissing values:")
print(df.isnull().sum())
print(f"\nData types:")
print(df.dtypes)
print(f"\nAccident_Severity unique values: {df['Accident_Severity'].unique().tolist()}")
print(f"\nTimestamp range: {df['Timestamp'].min()} → {df['Timestamp'].max()}")
print(f"\nSample cleaned data:")
print(df.head(3).to_string())
print()
```

```python
output_path = 'transportation_cleaned.csv'
df.to_csv(output_path, index=False)
print(f"Saved to: {output_path}")
print(f"Final shape: {df.shape[0]:,} rows × {df.shape[1]} columns")
print()
```

```python
import sqlite3

conn = sqlite3.connect('transportation.db')

df = pd.read_csv('transportation_cleaned.csv')
df.to_sql('transportation', conn, if_exists='replace', index=False)

print("Ready")
```
